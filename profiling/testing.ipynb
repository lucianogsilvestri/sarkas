```python
# Import the usual libraries
import numpy as np
import matplotlib.pyplot as plt 
%matplotlib inline
import os

plt.style.use('MSUstyle')

# Import sarkas
from sarkas.processes import PreProcess, Simulation, PostProcess
```

```python
# Create the file path to the YAML input file
N = 10000
input_file_name = os.path.join('input_files', 'BIM_cgs.yaml')
args = {'Particles': [
        {"Species" :{
            'name': 'H',
            'number_density': 8.1e+27,
            'atomic_weight': 1.0,           
            'Z': 1.0,
            'temperature_eV': 14.68,
            'num': N//2,
            'replace': True
            }
        },
        {"Species" :{
            'name' :'He',
            'number_density': 8.1e+27,
            'atomic_weight': 4,
            'Z': 2.0,
            'temperature_eV': 14.68,
            'num': N//2,
            'replace': True
            }
        },
    ],
    "IO" : {
                'verbose': True,
            'job_dir': f'N{N}'
                }
    }
pre = PreProcess(input_file_name)
pre.setup(read_yaml=True, other_inputs=args)
```

```python
pre.run(pppm_estimate=True)
```

```python
f"rc = {5.19*pre.parameters.a_ws:.6e}, alpha = {0.64 / pre.parameters.a_ws:.6e}"
```
